```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('https://stepik.org/media/attachments/lesson/16462/boston_houses.csv')
```

```python
arr = df.values
```

```python
x = arr[:, 1:]
y = arr[:, 0]
x = np.hstack((np.ones((x.shape[0], 1)), x))
w = np.linalg.inv(x.T @ x) @ x.T @ y

print(', '.join(map(str, w)))
```
